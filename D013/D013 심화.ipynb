# 집단 차이 검정 보고서

## 1. 질문과 데이터

- 결제 금액(tips)의 성별·시간대 차이 / 펭귄 체중(penguins)의 종별 차이

## 2. 결과 요약 (3종 세트)

| 비교            | 평균차 | 95% CI | 효과크기      | p    | 판정                |
| --------------- | ------ | ------ | ------------- | ---- | ------------------- |
| 성별(남-여)     | 2.69   | 0.45, 4.92   | d=0.30 (작음) | 0.0186 | 유의                |
| 시간(저녁-점심) | 3.63   | 1.33, 5.93   | d=0.41 (작음)  | 0.0022 | 유의                |

## 3. 의사결정 제안

- 성별 차이는 유의하나, 효과가 작아 단독으로 타깃을 분리하는 근거로는 약함 → 보류 
- 시간대 차이는 성별에 비해 효과는 더 크지만 여전히 해당 결과만으로 우선 실행보다는 추가 분석이나 A/B 테스트가 필요함 -> 보류 
- (가정 점검) penguins는 등분산이 깨져 Welch ANOVA/비모수 재확인 권장
