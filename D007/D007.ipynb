# 모두마켓 월간 로그 — 도구 선택 보고서

## 1. 데이터 개요
- 파일: web_logs.csv
- 크기: 32.3 MB, 500,000 행
- 주요 컬럼: ts / user_id / page / device / status_code / response_ms / bytes_sent

## 2. 측정 결과
| 방식 | 소요 시간 | 메모리 피크 |
|---|---|---|
| pandas + dtype | 2.30초 | 63.0MB |
| pandas chunked | 2.26초 | 17.4MB |
| Polars lazy    | 0.31초 | 7.1MB |

## 3. 분석 결과 요약
- 페이지별 평균 응답 시간 최고: mypage (132.0ms)
- 디바이스 점유율: mobile 69.9%, desktop 25.1%, tablet 4.9%
- 에러 1위 페이지: home (42898건)

## 4. 도구 선택 정당화 (한 단락)
이 분석에는 **Polaes lazy** 을 선택했습니다. 이유는
(1) 측정 결과 Polars lazy의 처리 시간이 0.31초로 가장 짧고 메모리 피크도 7.1MB로 가장 낮았으며,, (2) 팀 환경상 반복적인 로그 분석에서 처리 속도와 메모리 효율성이 중요하고, (3) lazy 실행 방식을 통해 필터링과 집계 과정을 최적화할 수 있어 향후 데이터 증가에도 유연하게 대응할 수 있기 때문입니다.

## 5. 다음 단계 제안
- 데이터 크기가 1억행 까지 늘어나면 Parquet 외부 도구의 도입을 고려.
- 시각화·검증은 D+008에서 Seaborn을 활용해 진행.

## 필수 과제

☑ web_logs.csv 를 pandas + dtype, 청크, Polars lazy 세 가지 방식으로 모두 분석했다.

☑ 세 가지 방식의 소요 시간과 메모리 사용량 을 측정해 표로 정리했다.

☑ 보고서의 "도구 선택 정당화" 한 단락을 측정 결과 + 팀 환경 두 근거로 작성했다.

☑ 분석 결과 3종(페이지별 응답시간 / 디바이스 점유율 / 에러 카운트)을 모두 표 또는 그림으로 정리했다.

## 심화 과제 (선택)

☑ chunksize 를 10K · 50K · 100K · 200K 로 바꿔 시나리오 2를 다시 측정하고, 청크 크기와 성능의 관계를 한 줄로 결론지었다.

X Polars lazy 의 .explain() 출력을 캡쳐해 어떤 최적화가 적용됐는지 한 줄로 설명했다.

X 동료가 모두 pandas만 쓰는 팀에 Polars 도입을 제안한다고 가정하고, 제안서 1단락 을 작성했다 (학습 비용·기대 효과·도입 단계).
