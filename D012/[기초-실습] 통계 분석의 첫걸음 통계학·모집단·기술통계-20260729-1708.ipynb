```python
# 데이터 생성
np.random.seed(2025)  # 재현 가능성을 위해 시드 고정

# 1반: 평균 75, 표준편차 8
class1 = np.random.normal(75, 8, 30)

# 2반: 일반 학생 + 이상값(0점, 130점)
class2 = np.concatenate([
    np.random.normal(75, 15, 28),  # 일반 학생
    [0, 130]                       # 극단적인 이상값
])

df3 = pd.DataFrame({
    'class': ['1반'] * 30 + ['2반'] * 30,
    'score': np.concatenate([class1, class2])
})

print(df3)
```

**📌 아래를 수행해 보세요:**

3. 이상값이 통계 지표에 어떤 영향을 주는지 생각해 봅시다.

```python
# 아래를 수행해 보세요:

score_mean1 = np.mean(class1)
score_mean2 = np.mean(class2)

score_median1 = np.median(class1) 
score_median2 = np.median(class2)

score_min1 = class1.min()
score_min2 = class2.min()

score_max1 = class1.max()
score_max2 = class2.max()

print("1반 평균 점수:", score_mean1)
print("2반 평균 점수:", score_mean2)

print("1반 중앙 점수:", score_median1)
print("2반 중앙 점수:", score_median2)

print("1반 최소 점수:", score_min1)
print("2반 최소 점수:", score_min2)

print("1반 최대 점수:", score_max1)
print("2반 최대 점수:", score_max2)

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df3,
    x="class",
    y="score"
)

plt.title("반별 점수 분포")
plt.xlabel("반")
plt.ylabel("점수")
plt.show()
```

# 문제 4. IQR 이상값 제거

**📘 문제**

- 앞의 문제에서 2반에는 **극단적인 이상값(예: 매우 낮은 점수)**이 포함되어 있어
  통계 지표, 특히 평균에 큰 영향을 주는 것을 확인했습니다.
- 이번에는 **IQR(사분위 범위)**을 기준으로 이상값을 제거한 뒤, 이상값 제거 전후의 평균을 비교해 봅시다

<br>

**📌 아래를 수행해 보세요:**

- 2반의 점수 데이터를 대상으로 IQR 기준을 사용해 이상값을 판별하고 제거해 봅시다.
- 이상값 제거 전과 제거 후의 평균을 각각 계산해 봅시다.
- 이상값 제거가 평균에 어떤 영향을 주는지 설명해 봅시다.
- 필요하다면 중앙값과도 비교해 봅시다.

```python
# 아래를 수행해 보세요:

clean_data = []
outlier_data = []

for class_name, group in df3.groupby("class"):

    Q1 = group["score"].quantile(0.25)
    Q3 = group["score"].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # 정상 범위의 데이터
    group_clean = group[
        (group["score"] >= lower_bound) &
        (group["score"] <= upper_bound)
    ]

    # 이상치 데이터
    group_outliers = group[
        (group["score"] < lower_bound) |
        (group["score"] > upper_bound)
    ]

    clean_data.append(group_clean)
    outlier_data.append(group_outliers)

# 반별 정상 데이터를 다시 하나의 데이터프레임으로 합침
df3_clean = pd.concat(clean_data, ignore_index=True)

# 제거된 이상치만 모음
df3_outliers = pd.concat(outlier_data, ignore_index=True)

print("이상치:")
print(df3_outliers)

print("이상치 제거 전:")
print(df3.groupby("class")["score"].mean())

print("\n이상치 제거 후:")
print(df3_clean.groupby("class")["score"].mean())
```

# 문제 9. 정규분포와 표준편차

**📘 문제**

- 정규분포와 표준편차의 개념을 실제 데이터에 적용해 보겠습니다.
- 정규분포를 따르는 점수 데이터에서 다음을 확인해 봅시다:
  - 평균 ±1σ, 평균 ±2σ 범위에 속한 값들의 비율은 실제로 얼마나 되는가?
  - 이 결과가 이론적인 68-95-99 법칙과 얼마나 유사한지 비교해 봅시다.

<br>

**📌 아래를 수행해 보세요:**

1. 평균이 70, 표준편차가 10인 정규분포 데이터를 1000개 생성해 봅시다.
2. 평균을 중심으로 ±1σ, ±2σ 범위에 포함되는 데이터의 비율을 계산해 봅시다.
3. 결과를 바탕으로 68-95-99 법칙과 비교해 봅시다.

```python
# 아래를 수행해 보세요:

np.random.seed(2025)

data = np.random.normal(
    loc=70,      # 정규분포의 평균
    scale=10,    # 정규분포의 표준편차
    size=1000
)

print(data[:10]) # numpy 배열은 head 대신 슬라이싱 활용

mean = np.mean(data)
std = np.std(data)

# 평균 ±1표준편차 범위에 포함되는지
within_1sigma = (
    (data >= mean - std) &
    (data <= mean + std)
)

# 평균 ±2표준편차 범위에 포함되는지
within_2sigma = (
    (data >= mean - 2 * std) &
    (data <= mean + 2 * std)
)

# True는 1, False는 0이므로 평균을 구하면 포함 비율이 됨
ratio_1sigma = within_1sigma.mean()
ratio_2sigma = within_2sigma.mean()

print("데이터 평균:", mean)
print("데이터 표준편차:", std)

print(f"±1σ 범위: {mean - std:.2f} ~ {mean + std:.2f}")
print(f"±1σ 포함 비율: {ratio_1sigma:.2%}")

print(f"±2σ 범위: {mean - 2 * std:.2f} ~ {mean + 2 * std:.2f}")
print(f"±2σ 포함 비율: {ratio_2sigma:.2%}")
```

**🧠 데이터를 어떻게 읽을까요?**

1. ±1σ, ±2σ 범위에 실제로 몇 %의 데이터가 포함되었나요?
2. 이 결과는 68–95–99 법칙과 얼마나 비슷했나요?
3. 표준편차는 왜 중요한 지표인가요?
4. 이 개념이 실제로 어떻게 활용될 수 있을까요?

```python
# 데이터를 어떻게 읽을까요?

# ±1σ, ±2σ 범위에 68-95-99 법칙과 거의 동일한 비율의 데이터가 포함됨
# 표준편차는 데이터가 평균을 중심으로 얼마나 퍼져 있는지를 나타내며, 데이터의 변동성과 일관성, 이상치 탐색 등에 활용됨.
# 가설검정에서는 표준오차와 검정통계량 계산에 사용되어 귀무가설의 기각 여부를 판단하는 데 중요한 역할을 함.
```

# 문제 11. 단순 선형 회귀 분석

**📘 문제**

- 공부 시간과 성적 간의 관계를 회귀 분석으로 확인해봅시다.

```python
# 데이터 생성
from sklearn.linear_model import LinearRegression

np.random.seed(505)
study_time = np.random.normal(5, 2, 100)
score = 50 + study_time * 8 + np.random.normal(0, 5, 100)
df11 = pd.DataFrame({'study_time': study_time, 'score': score})

print(df11)
```

**📌 아래를 수행해 보세요:**

1. 공부 시간(`study_time`)과 성적(`score`)간의 관계를 단순 선형 회귀모델로 분석합니다.
2. 회귀 모델을 학습하고, 기울기(계수)와 절편을 출력해 봅시다.
3. 산점도 위에 회귀선을 시각화해 봅시다.

```python
# 아래를 수행해 보세요:

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 데이터 생성
np.random.seed(505)

study_time = np.random.normal(5, 2, 100)
score = 50 + study_time * 8 + np.random.normal(0, 5, 100)

df11 = pd.DataFrame({
    "study_time": study_time,
    "score": score
})

# 1. 독립변수 X와 종속변수 y 설정
X = df11[["study_time"]]  # X는 2차원 형태로 입력
y = df11["score"]

# 2. 회귀 모델 생성 및 학습
model = LinearRegression()
model.fit(X, y)

# 3. 기울기와 절편 출력
slope = model.coef_[0]
intercept = model.intercept_

print(f"기울기(회귀계수): {slope:.2f}")
print(f"절편: {intercept:.2f}")
print(f"회귀식: score = {intercept:.2f} + {slope:.2f} × study_time")

# 4. 회귀선에 사용할 x값과 예측값 생성
x_line = np.linspace(
    df11["study_time"].min(),
    df11["study_time"].max(),
    100
).reshape(-1, 1)

y_pred = model.predict(x_line)

# 5. 산점도와 회귀선 시각화
plt.figure(figsize=(8, 5))

plt.scatter(
    df11["study_time"],
    df11["score"],
    label="실제 데이터"
)

plt.plot(
    x_line,
    y_pred,
    linewidth=2,
    label="회귀선"
)

plt.title("공부 시간과 성적의 단순 선형 회귀")
plt.xlabel("공부 시간")
plt.ylabel("성적")
plt.legend()
plt.show()
```

# 문제 12. A/B 테스트 해석

**📘 문제**

- 두 그룹의 전환율 차이를 계산하고 해석해봅시다.

```python
# 데이터 생성
np.random.seed(606)
group_a = np.random.binomial(1, 0.12, 1000)
group_b = np.random.binomial(1, 0.15, 1000)

print(group_a[:10])
print(group_b[:10])

# 전환율이란?
# 전체 대상 중에서 우리가 원하는 행동을 실제로 한 표본의 비율
```

3. 이 차이가 우연인지, 의미 있는 차이인지 생각해 봅시다.

```python
# 아래를 수행해 보세요:

# 0은 미전환, 1은 전환

# 전환 수
conversion_a = group_a.sum()
conversion_b = group_b.sum()

# 전체 사용자 수
n_a = len(group_a)
n_b = len(group_b)

# 전환율
conversion_rate_a = group_a.mean()
conversion_rate_b = group_b.mean()

# 전환율 차이: B - A
rate_difference = conversion_rate_b - conversion_rate_a

print(f"A 그룹 전환 수: {conversion_a}명 / {n_a}명")
print(f"B 그룹 전환 수: {conversion_b}명 / {n_b}명")

print(f"A 그룹 전환율: {conversion_rate_a:.2%}")
print(f"B 그룹 전환율: {conversion_rate_b:.2%}")

print(f"전환율 차이: {rate_difference:.2%}p")
```
